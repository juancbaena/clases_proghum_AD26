## Practicando con la tortuga
La tortuga tiene un "lápiz" que deja rastro. Imagina que está en una hoja de papel gigante. Cada comando es una instrucción precisa:

forward(X): Avanza X píxeles
backward(X): Retrocede X píxeles
left(X): Gira X grados a la izquierda
right(X): Gira X grados a la derecha
💻 Código de Ejemplo Completo:

```python
initializeTurtle()


shape("turtle")
speed(3)  # Velocidad: 1 (lento) a 10 (rápido), 0 es instantáneo

# Dibujar un cuadrado
for i in range(4):
    forward(100)
    left(90)

# Mover sin dibujar
penup()  # Levanta el lápiz
forward(150)  # Se mueve sin dejar rastro
pendown()  # Baja el lápiz

# Dibujar un triángulo
for i in range(3):
    forward(100)
    left(120)

#turtle.exitonclick() # To keep the window open until clicked (or in colab_turtle to show the final image)
```

## 🧮 Matemáticas de Ángulos

Fórmula mágica: Para polígonos regulares, el ángulo de giro es 360 / número_de_lados

Triángulo: 360/3 = 120°
Cuadrado: 360/4 = 90°
Pentágono: 360/5 = 72°
Hexágono: 360/6 = 60°

```python
#¿Qué crees que pase con este código?

initializeTurtle()

longitud_lado = 100
num_lados = 4
angulo = 360/num_lados


for i in range(num_lados):
  forward(longitud_lado)
  right(angulo)
```
