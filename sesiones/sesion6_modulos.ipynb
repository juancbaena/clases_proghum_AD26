## Módulos de Python
Un módulo en python es un programa (script) de Python que se instala usando el comando  `pip install` y nombre del módulo. Python tiene instalados por defecto algunos módulos, por ejemplo: math.
Para la manipulación de datos, los módulos más empleados son: pandas, numpy, matplotlib.
### Importar módulos en Python
Para usar un módulo en tu programa, primero debes importarlo. Hay varias maneras de importar módulos en Python:

Importación básica
Usa  `import` seguido del nombre del módulo. Por ejemplo:
 `import math `
Esto permite acceder a las funciones y constantes del módulo  `math `, como  `math.sqrt() ` para calcular raíces cuadradas.

````python
import math
````

````python
print(math.sqrt(25))
````

### Importar una función específica de un módulo
También puedes importar solo una función específica de un módulo, lo que es útil cuando necesitas una sola función y quieres reducir la cantidad de código. Ejemplo:

`from math import sqrt`
Ahora puedes usar `sqrt()` directamente en lugar de `math.sqrt()`.

````python
from math import sqrt

print(sqrt(25))
````

### Renombrar un módulo
Puedes asignar un nombre corto o alternativo a un módulo al importarlo. Esto es especialmente útil para módulos con nombres largos. Ejemplo:

`import numpy as np`
Aquí, puedes llamar a `numpy` simplemente como `np`, lo que facilita su uso.

````python
import numpy as np

coleccion = [3809, 12309, 1243213, 4543, 5823]

print(np.mean(coleccion))
````

### Librerías o módulos empleados para Machine Learning y para análisis de datos

---
* Numpy
    + NumPy es una librería esencial para el cálculo numérico en Python. Facilita el trabajo con matrices y operaciones matemáticas avanzadas, las cuales son básicas para machine learning. Uso principal: Manipulación de matrices y operaciones matemáticas.
* Pandas
    * pandas es una librería para manipulación y análisis de datos. Permite trabajar con estructuras de datos como DataFrames, facilitando el análisis de grandes volúmenes de datos en machine learning. Uso principal: Manipulación de datos estructurados
* Matplotlib
    + Matplotlib es una librería de gráficos que permite visualizar datos de manera clara. Es muy útil para representar resultados de modelos de machine learning. Uso principal: Visualización de datos.
* Scikit-learn
    + Scikit-learn es una librería que ofrece algoritmos de machine learning como clasificación, regresión y agrupamiento. Es una de las más usadas por su facilidad de uso y versatilidad. Uso principal: Aplicación de algoritmos de machine learning

````python
import numpy as np
import matplotlib.pyplot as plt

# Lista de visualizaciones de videos
visualizaciones = [1500, 2300, 3200, 4100, 2900]

# Calcular promedio y desviación estándar
promedio = np.mean(visualizaciones)
desviacion = np.std(visualizaciones)

# Mostrar resultados
print(f"El promedio de visualizaciones es: {promedio}")
print(f"La desviación estándar de las visualizaciones es: {desviacion}")

# Visualización de los datos
plt.bar(range(1, len(visualizaciones) + 1), visualizaciones)
plt.xlabel("Video (índice)")
plt.ylabel("Visualizaciones")
plt.title("Visualizaciones por Video")
plt.show()
````

## 1. Cargar la tabla con pandas

`pandas` puede leer un CSV **directamente desde una URL**. Lo guardamos en una variable llamada `df`
(de *DataFrame*, el nombre que pandas da a una tabla).


````python
import pandas as pd

url = "https://raw.githubusercontent.com/juancbaena/clases_proghum_AD26/main/otros/obras_literarias.csv"
df = pd.read_csv(url)

df.head()   # muestra las primeras filas
````

## 2. Explorar la tabla

Antes de analizar, conviene mirar el tamaño y las columnas.


````python
print("Filas y columnas:", df.shape)
print("Columnas:", list(df.columns))
print("\nTipos de dato por columna:")
print(df.dtypes)
````

Para ver una sola columna, se usa `df["nombre_columna"]`:


````python
print(df["genero"])
````

Y para un resumen numérico rápido de la columna de años o páginas:


````python
print(df["anio"].describe())
print("\nAnio mas antiguo:", df["anio"].min())
print("Anio mas reciente:", df["anio"].max())
````

## 3. Filtrar filas

Podemos quedarnos solo con las filas que cumplen una condición. Por ejemplo, las obras **posteriores a 1950**:


````python
obras_recientes = df[df["anio"] > 1950]
print("Obras posteriores a 1950:", len(obras_recientes))
obras_recientes[["titulo", "autor", "anio"]].head(10)
````

## 4. Contar por categoría con `value_counts()`

¿Cuántas obras hay de cada género? Una sola línea lo resuelve:


````python
conteo_generos = df["genero"].value_counts()
print(conteo_generos)
````

## 5. Agrupar con `groupby()`

`groupby` permite calcular algo **por grupo**. Por ejemplo, el promedio de páginas por género:


````python
paginas_por_genero = df.groupby("genero")["paginas"].mean()
print(paginas_por_genero.round(1))
````

## 6. Primer gráfico: barras por género

Ahora visualizamos. Importamos `matplotlib` y dibujamos cuántas obras hay por género.


````python
import matplotlib.pyplot as plt

conteo_generos = df["genero"].value_counts()

plt.figure(figsize=(8, 4))
plt.bar(conteo_generos.index, conteo_generos.values, color="steelblue")
plt.title("Numero de obras por genero")
plt.xlabel("Genero")
plt.ylabel("Cantidad de obras")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
````

## 7. Segundo gráfico: histograma de años

Un **histograma** muestra cómo se reparten los años de publicación: ¿se concentran en alguna época?


````python
plt.figure(figsize=(8, 4))
plt.hist(df["anio"], bins=10, color="indianred", edgecolor="white")
plt.title("Distribucion de las obras por año de publicacion")
plt.xlabel("Año")
plt.ylabel("Numero de obras")
plt.tight_layout()
plt.show()
````

> 🧭 **Nota — leer un gráfico con sentido crítico.** Un gráfico **no es neutral**:
> el número de barras del histograma (`bins`), la escala o los colores cambian la historia que cuenta.
> Una gran actividad para humanidades es pedir a los estudiantes que **interpreten** lo que ven
> ("¿por qué se concentran las obras en el siglo XX?") y que cuestionen las decisiones de diseño,
> en vez de tomar el gráfico como una verdad cerrada.
