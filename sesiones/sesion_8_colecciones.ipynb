# Listas

Las listas son como mochilas que pueden guardar muchos objetos. Cada objeto tiene una posición (índice) que empieza desde 0. Puedes acceder, modificar y agregar elementos.

```python
# Crear listas
videojuegos = ["Zelda", "Roblox", "Fortnite", "Mario", "Minecraft"]
numeros = [10, 20, 30, 40, 50]
mixta = ["texto", 42, True, 3.14]

# Acceder por índice
print(f"Primer juego: {videojuegos[0]}")
print(f"Tercer juego: {videojuegos[2]}")
print(f"Último juego: {videojuegos[-1]}")  # -1 es el último

# Longitud de la lista
print(f"Tienes {len(videojuegos)} videojuegos")

# Rebanadas (slices)
print(f"Primeros 3: {videojuegos[0:3]}")
print(f"Desde el índice 2: {videojuegos[2:]}")

# Verificar si existe
if "Minecraft" in videojuegos:
    print("¡Tienes Minecraft!")
```

# Modificación de listas

Las listas son dinámicas. Puedes agregar, eliminar, modificar y reordenar elementos en cualquier momento.

```python
# Lista inicial
mochila = ["libro", "lápiz", "botella"]
print(f"Mochila inicial: {mochila}")

# Agregar elementos
mochila.append("computadora")
print(f"Después de append: {mochila}")

# Insertar en posición específica
mochila.insert(1, "cuaderno")
print(f"Después de insert: {mochila}")

# Eliminar por valor
mochila.remove("lápiz")
print(f"Después de remove: {mochila}")

# Eliminar por índice
elemento_eliminado = mochila.pop(0)
print(f"Eliminé: {elemento_eliminado}")
print(f"Mochila actual: {mochila}")

# Modificar elemento
mochila[0] = "tablet"
print(f"Después de modificar: {mochila}")

# Ordenar
numeros = [5, 2, 8, 1, 9, 3]
numeros.sort()
print(f"Números ordenados: {numeros}")

# Invertir
numeros.reverse()
print(f"Números invertidos: {numeros}")
```

# Reto
Adivina el número secreto

```python
import random

print("🎯 ADIVINA EL NÚMERO SECRETO 🎯")
print("=" * 50)

# Configuración
numero_secreto = random.randint(1, 100)
intentos = 0
max_intentos = 7
adivinado = False

print(f"Tienes {max_intentos} intentos para adivinar un número del 1 al 100")
print("Te daré pistas: MÁS ALTO o MÁS BAJO\n")

# Bucle principal
while intentos < max_intentos and not adivinado:
    intentos += 1
    print(f"\n--- Intento {intentos}/{max_intentos} ---")
    
    try:
        intento = int(input("Tu predicción: "))
    except ValueError:
        print("⚠️ Debes escribir un número")
        intentos -= 1  # No contar intentos inválidos
        continue
    
    # Validar rango
    if intento < 1 or intento > 100:
        print("⚠️ El número debe estar entre 1 y 100")
        intentos -= 1
        continue
    
    # Evaluar
    if intento == numero_secreto:
        adivinado = True
        print(f"\n🎉 ¡CORRECTO! El número era {numero_secreto}")
        print(f"Lo adivinaste en {intentos} intentos")
    elif intento < numero_secreto:
        print("📈 El número secreto es MÁS ALTO")
    else:
        print("📉 El número secreto es MÁS BAJO")

# Si no adivinó
if not adivinado:
    print(f"\n💀 GAME OVER")
    print(f"El número secreto era: {numero_secreto}")

# Puntuación
print("\n" + "=" * 50)
if adivinado:
    if intentos <= 3:
        print("🏆 ¡INCREÍBLE! Eres un genio")
    elif intentos <= 5:
        print("⭐ ¡MUY BIEN! Lo lograste")
    else:
        print("👍 ¡BIEN! Lo conseguiste")
```

# Ejercicios con la tortuga
"Imagina que eres el Director Creativo de un sello discográfico independiente. Has recibido los archivos de la nueva banda 'The Tuples'. Tu tarea no es solo escuchar la música, sino visualizar su esencia.
Tienes 4 fuentes de datos (nuestras estructuras de datos) y debes usarlas para programar una 'Tortuga' que dibuje el logo oficial del álbum basándose en la complejidad y el estilo de la banda."

```python
!pip3 install ColabTurtle
from ColabTurtle.Turtle import *
```

```python
initializeTurtle()
# ==========================================
# BASE DE DATOS DE LA BANDA "THE TUPLES"
# ==========================================

# 1. LISTA (Lista de reproducción / Setlist)
# Las listas son ordenadas y permiten repetidos. 
# Usaremos esto para definir la "duración" o cantidad de trazos.
setlist = ["Intro", "Verse", "Chorus", "Solo", "Outro"]

# 2. TUPLA (Paleta de Colores RGB)
# Las tuplas son inmutables. Un color RGB no cambia una vez definido.
# Definimos 3 colores fijos para el diseño.
paleta_colores = ((255, 105, 180), (100, 149, 237), (32, 178, 170)) # Rosa, Azul, Turquesa

# 3. SET (Set de Instrumentos Únicos)
# Los sets eliminan duplicados. 
# Si el bajista toca 3 canciones, solo cuenta como 1 instrumento único.
# Esto definirá la "simetría" del dibujo.
instrumentos = {"Guitarra", "Bajo", "Batería", "Sintetizador", "Guitarra", "Voz"}

# 4. DICCIONARIO (Metadata del Álbum)
# Pares Clave-Valor. Usaremos esto para configurar la "tortuga".
album_info = {
    "nombre": "Data Structures",
    "bpm": 120,       # Velocidad de la tortuga
    "energia": 8,     # Grosor del lápiz (1-10)
    "anio": 2026
}

# ==========================================
# LÓGICA DE DIBUJO (LA TORTUGA)
# ==========================================

# Configuración inicial basada en el DICCIONARIO
#setup(600, 600) # Tamaño del lienzo
bgcolor("black") # Fondo negro (estilo consola/estudio)
speed(int(album_info["bpm"] / 10)) # Mapeamos BPM a velocidad
pensize(album_info["energia"]) # El grosor depende de la "energía"

# Cálculo matemático basado en el SET
# ¿Cuántos instrumentos únicos hay? Eso define los ángulos del polígono.
num_instrumentos = len(instrumentos) 
angulo_giro = 360 / num_instrumentos

print(f"Analizando {num_instrumentos} instrumentos únicos...")
print(f"Generando portada para: {album_info['nombre']}")

# ==========================================
# EL BUCLE CREATIVO
# ==========================================

# Iteramos sobre la LISTA (setlist)
for cancion in setlist:
    
    # Elegimos un color de la TUPLA (usando el índice de la canción)
    # Usamos módulo (%) para asegurar que el índice no se salga de la tupla
    indice_color = setlist.index(cancion) % len(paleta_colores)
    color_r, color_g, color_b = paleta_colores[indice_color]
    
    pencolor(color_r, color_g, color_b)
    
    # Dibujamos una forma geométrica basada en los instrumentos (SET)
    for i in range(num_instrumentos):
        forward(100)
        right(angulo_giro)
    
    # Al terminar cada "canción", rotamos un poco para crear un patrón espiral
    right(15)
```

# Explicando el código
* **La Lista (setlist)**:

    * **Concepto**: Secuencia y Orden.

        * **Explicación**: En un guion de cine o una línea de tiempo musical, el orden importa. No es lo mismo el intro que el outro. Por eso usamos Lista. Además, nos permitió iterar (hacer un bucle) sobre cada canción.

* **La Tupla (paleta_colores)**:

    * **Concepto**: Integridad de Datos.

        * Explicación: Un código de color HEX o RGB es una coordenada fija. No quieres que el programa cambie accidentalmente el "Rojo" a "Verde" durante la ejecución. Las tuplas protegen esa información de ser modificada. Analogía musical: Las coordenadas de un sample o la frecuencia de una nota.

* **El Set (instrumentos)**:

    * **Concepto**: Unicidad y Filtrado.

        * **Explicación**: En comunicación, cuando analizamos "tags" de una noticia o "hashtags", no nos importa si se repiten, nos importa cuántos temas únicos hay. El Set nos dio automáticamente el número de "caras" del polígono sin que tuviéramos que contar manualmente.

* **El Diccionario (album_info)**:

    * **Concepto**: Metadata y Estructura.

        * **Explicación**: Así como un archivo MP3 tiene ID3 tags (artista, año, álbum), nosotros creamos un objeto con propiedades. Nos permitió acceder a datos por nombre (album_info["bpm"]) en lugar de por posición, haciendo el código más legible y "humano".
