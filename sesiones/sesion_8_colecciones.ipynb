# Listas

Las listas son como mochilas que pueden guardar muchos objetos. Cada objeto tiene una posición (índice) que empieza desde 0. Puedes acceder, modificar y agregar elementos.

```python
# Crear listas
videojuegos = ["Zelda", "Roblox", "Fortnite", "Mario", "Minecraft"]
numeros = [10, 20, 30, 40, 50]
mixta = ["texto", 42, True, 3.14]

# Acceder por índice
print(f"Primer juego: {videojuegos[0]}")
print(f"Tercer juego: {videojuegos[2]}")
print(f"Último juego: {videojuegos[-1]}")  # -1 es el último

# Longitud de la lista
print(f"Tienes {len(videojuegos)} videojuegos")

# Rebanadas (slices)
print(f"Primeros 3: {videojuegos[0:3]}")
print(f"Desde el índice 2: {videojuegos[2:]}")

# Verificar si existe
if "Minecraft" in videojuegos:
    print("¡Tienes Minecraft!")
```

# Modificación de listas

Las listas son dinámicas. Puedes agregar, eliminar, modificar y reordenar elementos en cualquier momento.

```python
# Lista inicial
mochila = ["libro", "lápiz", "botella"]
print(f"Mochila inicial: {mochila}")

# Agregar elementos
mochila.append("computadora")
print(f"Después de append: {mochila}")

# Insertar en posición específica
mochila.insert(1, "cuaderno")
print(f"Después de insert: {mochila}")

# Eliminar por valor
mochila.remove("lápiz")
print(f"Después de remove: {mochila}")

# Eliminar por índice
elemento_eliminado = mochila.pop(0)
print(f"Eliminé: {elemento_eliminado}")
print(f"Mochila actual: {mochila}")

# Modificar elemento
mochila[0] = "tablet"
print(f"Después de modificar: {mochila}")

# Ordenar
numeros = [5, 2, 8, 1, 9, 3]
numeros.sort()
print(f"Números ordenados: {numeros}")

# Invertir
numeros.reverse()
print(f"Números invertidos: {numeros}")
```

# Ejercicios con la tortuga
"Imagina que eres el Director Creativo de un sello discográfico independiente. Has recibido los archivos de la nueva banda 'The Tuples'. Tu tarea no es solo escuchar la música, sino visualizar su esencia.
Tienes 4 fuentes de datos (nuestras estructuras de datos) y debes usarlas para programar una 'Tortuga' que dibuje el logo oficial del álbum basándose en la complejidad y el estilo de la banda."

```python
!pip3 install ColabTurtle
import random
from ColabTurtle.Turtle import *
```

```python

initializeTurtle()

print("GENERADOR DE PORTADAS 'THE TUPLES'")
print("Selecciona el 'Vibe' del álbum para configurar la tortuga:")
print("1. Cyberpunk (Rápido, Neón, Geometría caótica)")
print("2. Acústico (Lento, Tonos tierra, Espirales suaves)")
print("3. Metal (Pesado, Rojo/Negro, Polígonos complejos)")

vibe = input("Escribe 1, 2 o 3: ")

# 1. DICCIONARIO DE CONFIGURACIÓN (Metadata del Vibe)
config_vibes = {
    "1": {"color": (0, 255, 255), "velocidad": 13, "grosor": 3, "nombre": "Cyberpunk"},
    "2": {"color": (139, 69, 19), "velocidad": 3, "grosor": 8, "nombre": "Acústico"},
    "3": {"color": (255, 0, 0), "velocidad": 10, "grosor": 12, "nombre": "Metal"}
}

# Validación interactiva
while vibe not in config_vibes:
    vibe = input("Opción no válida. Escribe 1, 2 o 3: ")

config = config_vibes[vibe]

# 2. SET INTERACTIVO (El usuario decide la complejidad)
try:
    num_cuerdas = int(input(f"¿Cuántas 'cuerdas' o lados tendrá el polígono base? (Recomendado 3-12): "))
    angulo_giro = 360 / num_cuerdas
except ValueError:
    print("Usando 5 lados por defecto...")
    num_cuerdas = 5
    angulo_giro = 360 / 5

# 3. TUPLA DE COLORES (Paleta inmutable)
# Si es Cyberpunk, agregamos un color extra aleatorio al final
paleta_base = (config["color"], (255, 20, 147), (255, 255, 0)) 
if vibe == "1":
    paleta_base = paleta_base + ((0, 255, 0),) # Agregamos verde neón

# Configuración de la Tortuga
bgcolor("black")
speed(config["velocidad"])
pensize(config["grosor"])
penup()
goto((random.randint(0,800)), (random.randint(0,500)))
pendown()

print(f"\nRenderizando portada estilo: {config['nombre']}...")
print(f"Complejidad: {num_cuerdas} lados")

# 4. LISTA (El Setlist de trazos)
setlist = ["Intro", "Verse", "Chorus", "Solo", "Outro"]

for i, cancion in enumerate(setlist):
    # Elegir color de la TUPLA
    r, g, b = paleta_base[i % len(paleta_base)]
    pencolor(r, g, b)
    
    # Efecto "Glitch" aleatorio (Solo para Cyberpunk)
    if vibe == "1" and random.random() > 0.7:
        right(random.randint(-45, 45)) # Glitch de ángulo
        
    # Dibujar polígono basado en el SET/INPUT
    for _ in range(num_cuerdas):
        forward(50 + (i * 15)) # La LISTA (índice) hace que crezca en espiral
        right(angulo_giro)
        
    right(360 / len(setlist)) # Rotación por canción

print("¡Portada generada!")
```
